###  Definição da Função $a(n)$

De acordo com a lista, a sequência atribuída à minha matrícula foi a **2(c)**. O termo geral desta sequência é definido matematicamente por:

$$ a_n = \frac{\ln(n)}{n} $$

01. Escreva um programa e envie ao arquivo fonte ou um link do
mesmo, o qual tenha como ponto de partida uma funçai a(n) que
define o termo geral de uma sequˆencia numérica a = a(n) (use n

```python
def a(n):
  # Nesse caso, o n precisa ser um número pelo menos 1

  return np.log(n) / n
```

```python
print("=" *60)
print("Execução do Exercício 1")
print("=" *60)

n_min = int(input("Digite um valor mínimo(ex = 1): " ))
n_max = int(input("Digite um valor máximo(ex = 200): " ))


print("\n")

print("\nEscolha uma opção de execução:")
print("(a) Não sei se a sequência é convergente.")
print("(b) Sei que a sequência é convergente para um limite L.")
opcao = input("Opção (a/b): ").strip().lower()


L = None
eps = None
N_eps = None


if opcao == 'b':
    L = float(input("Digite o valor do Limite L (ex: 0): "))
    eps = float(input("Digite a tolerância epsilon (ex: 0.05): "))
    N_eps = int(input("Digite o índice N(epsilon): "))
```

#### Nessa parte são feitos os cálculos e a tabela

```python
vetor_tamanho_n = np.arange(n_min, n_max + 1)

vetor_a = a(vetor_tamanho_n)

tabela = pd.DataFrame({"n":vetor_tamanho_n, "a(n)":vetor_a})

print("*"*60)
print("Mostrando tabela de valores")
print("*"*60)

display(tabela)


```

#### 2.1 Plotagem do gráfico

```python

plt.figure(figsize=(10, 6))

# Item (ii): Gráfico discreto. 'scatter' plota pontos isolados em vez de uma linha contínua
plt.scatter(vetor_tamanho_n, vetor_a, color='blue', label='a(n)', zorder=3)

# Item (iii): Adicionando as linhas se a opção for 'b'
if opcao == 'b':
    # axhline desenha uma linha horizontal atravessando todo o gráfico
    plt.axhline(y=L, color='green', linestyle='-', label=f'Limite L = {L}')
    plt.axhline(y=L + eps, color='red', linestyle='--', label=f'L + eps = {L+eps}')
    plt.axhline(y=L - eps, color='red', linestyle='--', label=f'L - eps = {L-eps}')

plt.title('Comportamento Assintótico da Sequência a(n) = ln(n)/n')
plt.xlabel('Índice n')
plt.ylabel('Valor de a(n)')
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend() 


plt.show()
```
